```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')
```

```python
df
```

```python
cidade_residencia = df[(df['nome_municipio'] == 'Santa Rosa')]
cidade_residencia
```

```python
MESES = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
colunas_temp_media = [f'temp_media_{mes}' for mes in MESES]

temperaturas_por_mes = cidade_residencia[colunas_temp_media].iloc[0]
temperaturas_por_mes.index = MESES

mes_mais_frio = temperaturas_por_mes.idxmin()
menor_temperatura = temperaturas_por_mes.min()

print(f'mes com a menor temperatura media foi {mes_mais_frio} com {menor_temperatura:.2f} °C')
```

```python
colunas_deficit = [f'def_{mes}' for mes in MESES]

deficit_por_mes = cidade_residencia[colunas_deficit].iloc[0]
deficit_por_mes.index = MESES

mes_maior_deficit = deficit_por_mes.idxmax()
maior_deficit = deficit_por_mes.max()

print(f'mes com o maior deficit hidrico foi {mes_maior_deficit} com {maior_deficit:.2f} mm')
```

```python
ESTACAO_DO_MES = {
    'dez': 'Verão', 'jan': 'Verão', 'fev': 'Verão',
    'mar': 'Outono', 'abr': 'Outono', 'mai': 'Outono',
    'jun': 'Inverno', 'jul': 'Inverno', 'ago': 'Inverno',
    'set': 'Primavera', 'out': 'Primavera', 'nov': 'Primavera',
}

media_por_estacao = temperaturas_por_mes.groupby(ESTACAO_DO_MES, sort=False).mean()

print(media_por_estacao.round(2))
```
