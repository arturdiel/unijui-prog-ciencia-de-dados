```python
class Calculadora:

  def __init__(self):
    self.__historico = []

  def somar(self, a, b):
    resultado = a + b
    self.__historico.append(f"{a} + {b} = {resultado}")
    return resultado

  def subtrair(self, a, b):
    resultado = a - b
    self.__historico.append(f"{a} - {b} = {resultado}")
    return resultado

  def multiplicar(self, a, b):
    resultado = a * b
    self.__historico.append(f"{a} * {b} = {resultado}")
    return resultado

  def dividir(self, a, b):
    resultado = a / b
    self.__historico.append(f"{a} / {b} = {resultado}")
    return resultado

  def raizQuadrada(self, valor):
    resultado = self.__metodoNewton(valor)
    self.__historico.append(f"Raiz de {valor} = {resultado}")
    return resultado

  def exibirHistorico(self):
    if len(self.__historico) == 0:
      print("Nenhum calculo realizado.")
    else:
      for calculo in self.__historico:
        print(calculo)

  def exibirMenu(self):
    print("Informe um valor para selecionar a operacao:")
    print("1 - Somar")
    print("2 - Subtrair")
    print("3 - Multiplicar")
    print("4 - Dividir")
    print("5 - Raiz quadrada")
    print("6 - Historico")
    print("-1 - Sair")

  def __metodoNewton(self, valor):
    if valor == 0:
      return 0.0

    estimativa = valor + 1
    novaEstimativa = (estimativa + valor / estimativa) / 2

    while novaEstimativa < estimativa:
      estimativa = novaEstimativa
      novaEstimativa = (estimativa + valor / estimativa) / 2

    return estimativa
```

```python
calc = Calculadora()
operacao = 0

while True:
  if operacao == -1:
    break

  calc.exibirMenu()
  operacao = int(input("Informe o valor desejado: "))

  if operacao == 1:
    a = float(input("Primeiro valor: "))
    b = float(input("Segundo valor: "))
    print(f"Resultado: {calc.somar(a, b)}")

  elif operacao == 2:
    a = float(input("Primeiro valor: "))
    b = float(input("Segundo valor: "))
    print(f"Resultado: {calc.subtrair(a, b)}")

  elif operacao == 3:
    a = float(input("Primeiro valor: "))
    b = float(input("Segundo valor: "))
    print(f"Resultado: {calc.multiplicar(a, b)}")

  elif operacao == 4:
    a = float(input("Primeiro valor: "))
    b = float(input("Segundo valor: "))
    while b == 0:
      print("Nao e possivel dividir por zero.")
      b = float(input("Segundo valor: "))
    print(f"Resultado: {calc.dividir(a, b)}")

  elif operacao == 5:
    valor = float(input("Valor: "))
    while valor < 0:
      print("Nao existe raiz quadrada de numero negativo.")
      valor = float(input("Valor: "))
    print(f"Resultado: {calc.raizQuadrada(valor)}")

  elif operacao == 6:
    calc.exibirHistorico()

  elif operacao == -1:
    print("Calculadora encerrada.")

  else:
    print("Operacao invalida.")
```
