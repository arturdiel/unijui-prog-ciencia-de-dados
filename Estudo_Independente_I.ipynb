```python
# Desafio I - Análise de notas de uma turma

NOTA_MINIMA = 6.0

def calcular_maior_nota(notas):
    maior = notas[0]
    for nota in notas:
        if nota > maior:
            maior = nota
    return maior

def calcular_menor_nota(notas):
    menor = notas[0]
    for nota in notas:
        if nota < menor:
            menor = nota
    return menor

def calcular_media_da_turma(notas):
    soma = 0
    for nota in notas:
        soma = soma + nota
    return soma / len(notas)

def contar_aprovados(notas, nota_minima=NOTA_MINIMA):
    aprovados = 0
    for nota in notas:
        if nota >= nota_minima:
            aprovados = aprovados + 1
    return aprovados

def contar_reprovados(notas, nota_minima=NOTA_MINIMA):
    return len(notas) - contar_aprovados(notas, nota_minima)

def ler_quantidade_de_estudantes():
    quantidade = int(input("Informe a quantidade de estudantes: "))
    while quantidade <= 0:
        print("A turma precisa ter pelo menos 1 estudante.")
        quantidade = int(input("Informe a quantidade de estudantes: "))
    return quantidade

def ler_nota(numero_do_estudante):
    nota = float(input(f"Nota do estudante {numero_do_estudante}: "))
    while nota < 0 or nota > 10:
        print("A nota deve estar entre 0 e 10.")
        nota = float(input(f"Nota do estudante {numero_do_estudante}: "))
    return nota

def ler_notas(quantidade):
    notas = []
    for estudante in range(1, quantidade + 1):
        notas.append(ler_nota(estudante))
    return notas

def exibir_relatorio_da_turma(notas, nota_minima=NOTA_MINIMA):
    if len(notas) == 0:
        print("Nenhuma nota foi informada.")
    else:
        print("RELATÓRIO DA TURMA")
        print(f"Notas informadas: {notas}")
        print(f"Maior nota: {calcular_maior_nota(notas):.1f}")
        print(f"Menor nota: {calcular_menor_nota(notas):.1f}")
        print(f"Média da turma: {calcular_media_da_turma(notas):.2f}")
        print(f"Aprovados: {contar_aprovados(notas, nota_minima)}")
        print(f"Reprovados: {contar_reprovados(notas, nota_minima)}")

def executar_desafio_1():
    quantidade = ler_quantidade_de_estudantes()
    notas = ler_notas(quantidade)
    exibir_relatorio_da_turma(notas)

executar_desafio_1()
```

```python
# Desafio II - Cadastro e consulta de produtos

NOME = 0
PRECO = 1
QUANTIDADE = 2

def cadastrar_produto(produtos, codigos, codigo, nome, preco, quantidade):
    if codigo in codigos:
        return False
    produtos[codigo] = [nome, preco, quantidade]
    codigos.append(codigo)
    return True

def consultar_produto(produtos, codigos, codigo):
    if codigo in codigos:
        return produtos[codigo]
    return []

def calcular_valor_total_estoque(produtos, codigos):
    total = 0
    for codigo in codigos:
        produto = produtos[codigo]
        total = total + produto[PRECO] * produto[QUANTIDADE]
    return total

def encontrar_maior_valor_unitario(produtos, codigos):
    codigo_mais_caro = ""
    maior_preco = 0
    for codigo in codigos:
        if codigo_mais_caro == "" or produtos[codigo][PRECO] > maior_preco:
            maior_preco = produtos[codigo][PRECO]
            codigo_mais_caro = codigo
    return codigo_mais_caro

def exibir_produto(codigo, produto):
    print(f"Código: {codigo}")
    print(f"Nome: {produto[NOME]}")
    print(f"Preço: R$ {produto[PRECO]:.2f}")
    print(f"Quantidade: {produto[QUANTIDADE]}")
    print(f"Subtotal: R$ {produto[PRECO] * produto[QUANTIDADE]:.2f}")

def listar_produtos(produtos, codigos):
    if len(codigos) == 0:
        print("Nenhum produto cadastrado.")
    else:
        for codigo in codigos:
            produto = produtos[codigo]
            subtotal = produto[PRECO] * produto[QUANTIDADE]
            print(f"{codigo} | {produto[NOME]} | R$ {produto[PRECO]:.2f} | {produto[QUANTIDADE]} un. | Subtotal: R$ {subtotal:.2f}")

def exibir_menu():
    print("CADASTRO DE PRODUTOS")
    print("1 - Cadastrar produto")
    print("2 - Consultar produto pelo código")
    print("3 - Listar todos os produtos")
    print("4 - Calcular valor total do estoque")
    print("5 - Produto de maior valor unitário")
    print("0 - Sair")

def ler_novo_produto(produtos, codigos):
    codigo = input("Código: ")
    while codigo in codigos:
        print("Já existe um produto com esse código.")
        codigo = input("Código: ")

    nome = input("Nome: ")
    preco = float(input("Preço: R$ "))
    quantidade = int(input("Quantidade em estoque: "))

    cadastrar_produto(produtos, codigos, codigo, nome, preco, quantidade)
    print(f"Produto '{nome}' cadastrado com sucesso.")

def executar_desafio_2():
    produtos = {}
    codigos = []
    opcao = ""

    while opcao != "0":
        exibir_menu()
        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            ler_novo_produto(produtos, codigos)

        elif opcao == "2":
            codigo = input("Código do produto: ")
            produto = consultar_produto(produtos, codigos, codigo)
            if len(produto) == 0:
                print("Produto não encontrado.")
            else:
                exibir_produto(codigo, produto)

        elif opcao == "3":
            listar_produtos(produtos, codigos)

        elif opcao == "4":
            total = calcular_valor_total_estoque(produtos, codigos)
            print(f"Valor total do estoque: R$ {total:.2f}")

        elif opcao == "5":
            codigo = encontrar_maior_valor_unitario(produtos, codigos)
            if codigo == "":
                print("Nenhum produto cadastrado.")
            else:
                print("Produto de maior valor unitário:")
                exibir_produto(codigo, produtos[codigo])

        elif opcao == "0":
            print("Encerrando o programa.")

        else:
            print("Opção inválida. Escolha um número do menu.")

executar_desafio_2()
```

```python
# Desafio III - Sistema de Estatísticas

VALOR_ENCERRAMENTO = 0

def ler_numeros(valor_encerramento=VALOR_ENCERRAMENTO):
    numeros = []
    print(f"Digite os números. Para encerrar, digite {valor_encerramento}.")

    numero = float(input("Número: "))
    while numero != valor_encerramento:
        numeros.append(numero)
        numero = float(input("Número: "))

    return numeros

def contar_valores(numeros):
    return len(numeros)

def calcular_media(numeros):
    soma = 0
    for numero in numeros:
        soma = soma + numero
    return soma / len(numeros)

def calcular_mediana(numeros):

    ordenados = []
    for numero in numeros:
        ordenados.append(numero)
    ordenados.sort()

    meio = int(len(ordenados) / 2)
    if len(ordenados) % 2 == 1:
        return ordenados[meio]
    return (ordenados[meio - 1] + ordenados[meio]) / 2

def encontrar_maior(numeros):
    maior = numeros[0]
    for numero in numeros:
        if numero > maior:
            maior = numero
    return maior

def encontrar_menor(numeros):
    menor = numeros[0]
    for numero in numeros:
        if numero < menor:
            menor = numero
    return menor

def eh_inteiro(numero):
    return numero == int(numero)

def contar_pares(numeros):
    pares = 0
    for numero in numeros:
        if eh_inteiro(numero) and numero % 2 == 0:
            pares = pares + 1
    return pares

def contar_impares(numeros):
    impares = 0
    for numero in numeros:
        if eh_inteiro(numero) and numero % 2 != 0:
            impares = impares + 1
    return impares

def contar_decimais(numeros):
    decimais = 0
    for numero in numeros:
        if not eh_inteiro(numero):
            decimais = decimais + 1
    return decimais

def exibir_estatisticas(numeros):
    if len(numeros) == 0:
        print("Nenhum número foi informado.")
    else:
        print("ESTATÍSTICAS")
        print(f"Valores informados: {numeros}")
        print(f"Quantidade: {contar_valores(numeros)}")
        print(f"Média: {calcular_media(numeros):.2f}")
        print(f"Mediana: {calcular_mediana(numeros):.2f}")
        print(f"Maior valor: {encontrar_maior(numeros):.2f}")
        print(f"Menor valor: {encontrar_menor(numeros):.2f}")
        print(f"Números pares: {contar_pares(numeros)}")
        print(f"Números ímpares: {contar_impares(numeros)}")

        decimais = contar_decimais(numeros)
        if decimais > 0:
            print(f"Valores decimais: {decimais} (não são pares nem ímpares)")

def executar_desafio_3():
    numeros = ler_numeros()
    exibir_estatisticas(numeros)

executar_desafio_3()
```
