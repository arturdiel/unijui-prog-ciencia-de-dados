```python
def calcula_percento(valor, percentual=100):
  quanto = valor * percentual / 100
  return quanto

print("20% de 300 sao {:10.2f}".format(calcula_percento(300,20)))
print("100% de 350 sao {:10.2f}".format(calcula_percento(50)))
```

```python
import math

n = float(input("Digite o número: "))

if n < 0:
    print("Não existe raiz quadrada real de um número negativo.")
elif n == 0:
    print("A raiz quadrada de 0 é 0.")
else:
    b = n
    p = (b + (n / b)) / 2
    passo = 1

    while abs(n - (p * p)) >= 0.0001:
        b = p
        p = (b + (n / b)) / 2
        passo = passo + 1

    print()
    print(f"Método de Newton: {p}")
    print(f"Função sqrt(): {math.sqrt(n)}")
```
