```python
# Atividade 01

import random
import json

valorGerado = random.randint(0, 15)
valorChute = None
rodada = 0
qtdTentativas = 0
valorSaida = -1
jsonValue = []

print("Informe um valor entre 0 e 15")
print(f"Digite {valorSaida} para SAIR")

while True:
  valorChute = int(input("Informe um valor: "))

  if valorChute == valorSaida:
    break

  qtdTentativas += 1

  if valorChute == valorGerado:
    rodada += 1
    jsonValue.append({
        "rodada": rodada,
        "numeroGerado": valorGerado,
        "qtdTentativas": qtdTentativas
    })
    with open("sample_data/numero.json", "w") as json_file:
      json.dump(jsonValue, json_file)

    print(f"Você acertou, parabéns! O valor é {valorGerado}")
    valorGerado = random.randint(0, 15)
    qtdTentativas = 0

print("\nHistórico:")
for item in jsonValue:
  print(f"Rodada {item['rodada']}: número {item['numeroGerado']} em {item['qtdTentativas']} tentativa(s)")

print("\nEncerrando programa...")
```

```python
1# Atividade 02

import csv

lista = []
valorTotal = 0

while True:
  item = input("Item: ")

  if item == "":
    break

  qtdItem = int(input(f"Quantidade: "))
  valorItem = float(input(f"Valor unitário: "))
  subtotal = valorItem * qtdItem
  valorTotal += subtotal

  lista.append({
      "item": item,
      "quantidade": qtdItem,
      "valorUnitario": valorItem,
      "subtotal": subtotal
  })

  with open("sample_data/compras.csv", "w", newline="") as f:
    escritor = csv.writer(f)
    escritor.writerow(["item", "quantidade", "valorUnitario", "subtotal"])
    for item in lista:
      escritor.writerow([item["item"], item["quantidade"], item["valorUnitario"], item["subtotal"]])

print("\nLista:")
for item in lista:
  print(f"{item['item']}: R$ {item['valorUnitario']:.2f} x {item['quantidade']} = R$ {item['subtotal']:.2f}")

print(f"\nValor total: R$ {valorTotal:.2f}")

```
