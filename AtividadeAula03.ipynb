```python
nome = str(input("Informe seu nome: "))
a = int (input ("Informe um valor de A: "))
b = int (input ("Informe um valor de B: "))
c = int (input ("Informe um valor de C: "))

aFloat = float(a)
bFloat = float(b)
cFloat = float(c)

media = (aFloat+bFloat+cFloat)/3
print(f"O aluno {nome}, recebeu as seguintes notas {aFloat}, {bFloat} e {cFloat}. Com isso, sua média ficou {media}")
```
