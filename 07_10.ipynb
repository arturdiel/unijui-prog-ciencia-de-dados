```python
# Importando a biblioteca
import numpy as np

# Exibe os números sem notação científica e com 2 casas decimais
np.set_printoptions(suppress=True, precision=2)
```

```python
# Dados da semana: cada linha é um dia
dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array([
    "Segunda",
    "Terça",
    "Quarta",
    "Quinta",
    "Sexta",
    "Sábado",
    "Domingo"
])

# Nome de cada coluna de dados, na mesma ordem
indicadores = np.array([
    "Pedidos",
    "Faturamento",
    "Clientes",
    "Ticket_Medio",
    "Reclamações"
])
```

```python
# 1.1 - Estrutura do array
print("Quantidade de linhas:", dados.shape[0])
print("Quantidade de colunas:", dados.shape[1])
print("Quantidade total de valores:", dados.size)
print("Número de dimensões:", dados.ndim)
print("Tipo dos dados:", dados.dtype)
```

```python
# 2.1 - Dados da sexta-feira (linha de índice 4)
print("Sexta-feira:", dados[4])
```

```python
# 2.2 - Faturamento de todos os dias (coluna de índice 1)
faturamento = dados[:, 1]
print(faturamento)
```

```python
# 2.3 - Reclamações de todos os dias (coluna de índice 4)
reclamacoes = dados[:, 4]
print(reclamacoes)
```

```python
# 2.4 - De quinta-feira (índice 3) até o fim do array (domingo)
print(dias[3:])
print(dados[3:])
```

```python
# 2.5 - Colunas de pedidos, faturamento e clientes (índices 0, 1 e 2; o 3 não entra)
print(dados[:, 0:3])
```

```python
# Separando as demais colunas usadas nas próximas partes
pedidos = dados[:, 0]
clientes = dados[:, 2]
```

```python
# 3.1 - Total de pedidos da semana
print(f"Total de pedidos: {np.sum(pedidos):.0f}")

# 3.2 - Faturamento total
print(f"Faturamento total: R$ {np.sum(faturamento):.2f}")

# 3.3 - Total de clientes
print(f"Total de clientes: {np.sum(clientes):.0f}")

# 3.4 - Média diária de pedidos
print(f"Média diária de pedidos: {np.mean(pedidos):.2f}")

# 3.5 - Média diária de faturamento
print(f"Média diária de faturamento: R$ {np.mean(faturamento):.2f}")

# 3.6 - Média de reclamações por dia
print(f"Média de reclamações por dia: {np.mean(reclamacoes):.2f}")
```

```python
# 4.1 - Dias com maior e menor número de pedidos
# argmax e argmin devolvem o índice (posição), que é usado para buscar o nome em dias
indice_mais_pedidos = np.argmax(pedidos)
indice_menos_pedidos = np.argmin(pedidos)

print(f"Maior número de pedidos: {dias[indice_mais_pedidos]} ({pedidos[indice_mais_pedidos]:.0f} pedidos)")
print(f"Menor número de pedidos: {dias[indice_menos_pedidos]} ({pedidos[indice_menos_pedidos]:.0f} pedidos)")
```

```python
# 4.2 - Dias com maior e menor faturamento
indice_maior_faturamento = np.argmax(faturamento)
indice_menor_faturamento = np.argmin(faturamento)

print(f"Maior faturamento: {dias[indice_maior_faturamento]} (R$ {faturamento[indice_maior_faturamento]:.2f})")
print(f"Menor faturamento: {dias[indice_menor_faturamento]} (R$ {faturamento[indice_menor_faturamento]:.2f})")
```

```python
# 4.3 - Amplitude do faturamento (maior - menor)
amplitude = np.max(faturamento) - np.min(faturamento)
print(f"Amplitude do faturamento: R$ {amplitude:.2f}")
```

```python
# 5.1 - Dias com faturamento superior a R$ 20.000
faturamento_acima_20000 = faturamento > 20000
print(dias[faturamento_acima_20000])
```

```python
# 5.2 - Dias com mais de 150 pedidos
mais_de_150_pedidos = pedidos > 150
print(dias[mais_de_150_pedidos])
```

```python
# 5.3 - Dias com mais de 5 reclamações
mais_de_5_reclamacoes = reclamacoes > 5
print(dias[mais_de_5_reclamacoes])
```

```python
# 5.4 - Faturamento dos dias com mais de 150 pedidos
# (a máscara dos pedidos é aplicada no array do faturamento)
print(faturamento[mais_de_150_pedidos])
```

```python
# 5.5 - Dias com faturamento abaixo da média semanal
media_faturamento = np.mean(faturamento)
abaixo_da_media = faturamento < media_faturamento

print(f"Média semanal: R$ {media_faturamento:.2f}")
print("Dias abaixo da média:", dias[abaixo_da_media])
# np.sum conta os True, pois True vale 1 e False vale 0
print("Quantidade de dias:", np.sum(abaixo_da_media))
```

```python
# 6.1 - Classificação de cada dia
# np.where(condição, valor se True, valor se False)
classificacao = np.where(
    faturamento >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)

print(classificacao)

# junta dias e classificacao em uma matriz; o .T (transposta) deixa cada dia ao lado da sua classificação
print(np.array([dias, classificacao]).T)
```

```python
# 6.2 - Quantidade de dias de alto desempenho
alto_desempenho = classificacao == "Alto desempenho"
print("Dias de alto desempenho:", np.sum(alto_desempenho))
```

```python
# Medidas estatísticas do faturamento
media = np.mean(faturamento)
mediana = np.median(faturamento)
desvio_padrao = np.std(faturamento)
minimo = np.min(faturamento)
maximo = np.max(faturamento)

print(f"Média: R$ {media:.2f}")
print(f"Mediana: R$ {mediana:.2f}")
print(f"Desvio padrão: R$ {desvio_padrao:.2f}")
print(f"Mínimo: R$ {minimo:.2f}")
print(f"Máximo: R$ {maximo:.2f}")

# quanto o desvio padrão representa da média
print(f"Desvio padrão em relação à média: {desvio_padrao / media * 100:.2f}%")
```

```python
# Ordem dos indicadores nos resultados
print(indicadores)

# 8.1 - Média de cada indicador na semana
medias = np.mean(dados, axis=0)
print("Média:", medias)

# 8.2 - Maior valor de cada indicador
maiores = np.max(dados, axis=0)
print("Maior:", maiores)

# 8.3 - Menor valor de cada indicador
menores = np.min(dados, axis=0)
print("Menor:", menores)
```

```python
# 8.4 - Variação relativa: amplitude (maior - menor) dividida pela média
# a conta é feita para os 5 indicadores de uma vez
variacao_relativa = (maiores - menores) / medias
print(indicadores)
print("Variação relativa:", variacao_relativa)

indice_maior_variacao = np.argmax(variacao_relativa)
print("Maior variação relativa:", indicadores[indice_maior_variacao])
```

```python
# 9.1 - Melhor dia da semana pelo faturamento
indice_melhor_dia = np.argmax(faturamento)
print(f"Melhor dia: {dias[indice_melhor_dia]} (R$ {faturamento[indice_melhor_dia]:.2f})")
```

```python
# 9.2 - O melhor dia também teve a maior quantidade de pedidos?
indice_mais_pedidos = np.argmax(pedidos)
print("Dia com mais pedidos:", dias[indice_mais_pedidos])

if indice_melhor_dia == indice_mais_pedidos:
    print("Sim, o dia de maior faturamento também teve mais pedidos.")
else:
    print("Não, o dia de maior faturamento não foi o de mais pedidos.")
```

```python
# 9.3 - O dia de maior faturamento também teve mais reclamações?
indice_mais_reclamacoes = np.argmax(reclamacoes)
print("Dia com mais reclamações:", dias[indice_mais_reclamacoes])

if indice_melhor_dia == indice_mais_reclamacoes:
    print("Sim, o dia de maior faturamento também teve mais reclamações.")
else:
    print("Não, o dia de maior faturamento não foi o de mais reclamações.")
```

```python
# 9.4 - Dias com mais de 150 pedidos E faturamento acima de R$ 20.000
# o & exige que as duas condições sejam True no mesmo dia
pedidos_e_faturamento_altos = (pedidos > 150) & (faturamento > 20000)

print("Dias:", dias[pedidos_e_faturamento_altos])
print("Quantidade:", np.sum(pedidos_e_faturamento_altos))
```

```python
# Separando as colunas
pedidos = dados[:, 0]
faturamento = dados[:, 1]
reclamacoes = dados[:, 4]

# Índices do melhor dia, do pior dia e do dia com mais reclamações
indice_melhor_dia = np.argmax(faturamento)
indice_pior_dia = np.argmin(faturamento)
indice_mais_reclamacoes = np.argmax(reclamacoes)

# Dias com faturamento de R$ 20.000 ou mais
dias_alto_desempenho = dias[faturamento >= 20000]

# Exibindo o resumo
print("RESUMO DA SEMANA")
print(f"Total de pedidos: {np.sum(pedidos):.0f}")
print(f"Faturamento total: R$ {np.sum(faturamento):.2f}")
print(f"Média diária de pedidos: {np.mean(pedidos):.2f}")
print(f"Média diária de faturamento: R$ {np.mean(faturamento):.2f}")
print(f"Melhor dia: {dias[indice_melhor_dia]} (R$ {faturamento[indice_melhor_dia]:.2f})")
print(f"Pior dia: {dias[indice_pior_dia]} (R$ {faturamento[indice_pior_dia]:.2f})")
print(f"Maior número de reclamações: {reclamacoes[indice_mais_reclamacoes]:.0f} ({dias[indice_mais_reclamacoes]})")
print(f"Dias de alto desempenho: {dias_alto_desempenho.size} {dias_alto_desempenho}")
```

```python
# Reclamações a cada 100 pedidos (divide um array pelo outro, dia a dia)
reclamacoes_por_100_pedidos = reclamacoes / pedidos * 100
print(dias)
print(reclamacoes_por_100_pedidos)
```
