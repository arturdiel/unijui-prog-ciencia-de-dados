```python
import json
import requests
import pandas as pd

# local e periodo da consulta
latitude = -27.87
longitude = -54.48
data_inicial = "2021-01-01"
data_final = "2026-10-31"

url = (
    "https://archive-api.open-meteo.com/v1/archive"
    f"?latitude={latitude}&longitude={longitude}"
    f"&start_date={data_inicial}&end_date={data_final}"
    "&daily=temperature_2m_max,temperature_2m_min,precipitation_sum"
    "&timezone=America/Sao_Paulo"
)

resposta = requests.get(url)
print(resposta.status_code)
```

```python
dados = json.loads(resposta.text)

print("ponto usado pela API:", dados["latitude"], dados["longitude"])
print("altitude em metros:", dados["elevation"])
print("fuso horario:", dados["timezone"])
print("unidades:", dados["daily_units"])
```

```python
df = pd.DataFrame(dados["daily"])

df = df.rename(columns={
    "time": "data",
    "temperature_2m_max": "temp_max_c",
    "temperature_2m_min": "temp_min_c",
    "precipitation_sum": "chuva_mm"
})

df
```

```python
maior_temperatura = df["temp_max_c"].max()
maior_chuva = df["chuva_mm"].max()

print(f"quantidade de dias: {df.shape[0]}")
print(f"quantidade de colunas: {df.shape[1]}")
print(f"maior temperatura: {maior_temperatura} °C")
print(f"maior chuva em um dia: {maior_chuva} mm")
```

```python
df.to_csv("clima_santa_rosa.csv", index=False)
print("Arquivo salvo com sucesso!")

df_lido = pd.read_csv("clima_santa_rosa.csv")
df_lido
```
