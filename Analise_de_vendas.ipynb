```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dados = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"],
    "Notebook": [20, 25, 22, 30, 35, 40],
    "Mouse": [50, 55, 48, 60, 65, 70],
    "Teclado": [30, 28, 35, 40, 38, 45]
}

df = pd.DataFrame(dados)
print(df)

produtos = ["Notebook", "Mouse", "Teclado"]
```

```python
print("Primeiras linhas:")
print(df.head())

print("\nQuantidade de linhas (meses):", df.shape[0])
print("Quantidade de colunas:", df.shape[1])

print("\nEstatísticas descritivas:")
print(df.describe())
```

```python
for produto in produtos:
    print(f"{produto}: média {np.mean(df[produto]):.2f} | maior {np.max(df[produto])} | menor {np.min(df[produto])}")
```

```python
plt.plot(df["Mes"], df["Notebook"], label="Notebook")
plt.plot(df["Mes"], df["Mouse"], label="Mouse")
plt.plot(df["Mes"], df["Teclado"], label="Teclado")
plt.title("Vendas por mês")
plt.xlabel("Mês")
plt.ylabel("Unidades vendidas")
plt.legend()
plt.show()
```

```python
totais = df[produtos].sum()
print("Total vendido no período:")
print(totais)

plt.bar(totais.index, totais)
plt.title("Total vendido por produto (Jan a Jun)")
plt.xlabel("Produto")
plt.ylabel("Unidades vendidas")
plt.show()
```

```python
plt.hist(df["Mouse"], bins=3)
plt.title("Distribuição das vendas mensais do Mouse")
plt.xlabel("Unidades vendidas no mês")
plt.ylabel("Quantidade de meses")
plt.show()
```

```python
print(f"Produto com maior volume: {totais.idxmax()} ({totais.max()} unidades)")

for produto in produtos:
    indice = np.argmax(df[produto])
    print(f"{produto}: maior venda em {df['Mes'][indice]} ({df[produto][indice]} unidades)")
```

```python
# Produto com maior volume: Mouse, com 348 unidades.
# Mês de maior venda: junho para os três produtos (Notebook 40, Mouse 70 e Teclado 45).
#Conclusão: os gráficos mostram que as vendas dos três produtos cresceram de janeiro a junho, com o Mouse sempre na frente.
```

```python
media_notebook = np.mean(df["Notebook"])
print(f"Média de vendas do Notebook: {media_notebook:.2f}")

abaixo_da_media = df[df["Notebook"] < media_notebook]
print("\nMeses abaixo da média:")
print(abaixo_da_media[["Mes", "Notebook"]])

cores = np.where(df["Notebook"] < media_notebook, "red", "blue")

plt.bar(df["Mes"], df["Notebook"], color=cores)
plt.title(f"Notebook: vermelho = abaixo da média ({media_notebook:.2f})")
plt.xlabel("Mês")
plt.ylabel("Unidades vendidas")
plt.show()
```

```python
# Por que gráfico de barras: ele compara os meses lado a lado, e a cor vermelha destaca os meses abaixo da média (28,67).
# O que o gráfico mostra: janeiro, fevereiro e março ficaram abaixo da média, e abril, maio e junho acima. As vendas de Notebook melhoraram no segundo trimestre.
```
