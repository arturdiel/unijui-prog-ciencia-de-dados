```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
```

```python
navegador = gs.Chrome()
```

```python
url = "https://www.cptec.inpe.br/"
```

```python
navegador.get(url)
sleep(3)
print("Página:", navegador.title)
```

```python
botoes = navegador.find_elements(
    By.XPATH,
    "//button[@aria-label='Fechar tutorial']"
)
for botao in botoes:
    if botao.is_displayed():
        navegador.execute_script(
            "arguments[0].click();", botao
        )
        break
sleep(1)
```

```python
botoes = navegador.find_elements(
    By.XPATH,
    "//button[contains(., 'Não exibir novamente')]"
)
for botao in botoes:
    if botao.is_displayed():
        navegador.execute_script(
            "arguments[0].click();", botao
        )
        break
sleep(1)
```

```python
sugestoes = []
while len(sugestoes) == 0:
    cidade = input("Informe a cidade: ")

    caixa_busca = navegador.find_element(
        By.XPATH,
        "//input[@id='searchbox']"
    )
    caixa_busca.send_keys(cidade)
    sleep(3)

    sugestoes = navegador.find_elements(
        By.XPATH,
        "//ul[@id='autocomplete-results']/li"
    )

    if len(sugestoes) == 0:
        print("Cidade não encontrada.")
        navegador.get(url)
        sleep(3)
```

```python
print("\nCIDADES ENCONTRADAS:")
for numero in range(1, len(sugestoes) + 1):
    print(f"{numero} - {sugestoes[numero - 1].text}")

escolha = int(input("Informe o número da cidade: "))
while escolha < 1 or escolha > len(sugestoes):
    print("Número inválido.")
    escolha = int(input("Informe o número da cidade: "))

navegador.execute_script(
    "arguments[0].click();", sugestoes[escolha - 1]
)
sleep(4)
```

```python
cidade_encontrada = navegador.find_element(
    By.XPATH,
    "//span[@id='cidade_uf']"
).text

minima = navegador.find_element(
    By.XPATH,
    "//div[span[normalize-space()='Temp. Mín.']]/span[1]"
).text

maxima = navegador.find_element(
    By.XPATH,
    "//div[span[normalize-space()='Temp. Máx.']]/span[1]"
).text
```

```python
print("\nTEMPERATURA DO DIA:")
print("Cidade:", cidade_encontrada)
print("Mínima:", minima)
print("Máxima:", maxima)
```

```python
navegador.quit()
```
